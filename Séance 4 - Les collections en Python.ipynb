# Les collections en Python
## 1. Introduction

Jusqu'à présent, nous avons principalement manipulé des variables contenant une seule valeur :

```python
age = 25
nom = "Ali"
prix = 1500
```

Python propose plusieurs structures permettant de regrouper ces données. On les appelle des **collections**.

Dans ce cours, nous allons étudier trois collections fondamentales :

- la liste (list) ;
- le tuple (tuple) ;
- le dictionnaire (dict).

Important : dans ce chapitre, nous allons apprendre à créer, consulter et modifier ces collections sans utiliser de boucles. Les boucles seront utilisées dans un chapitre ultérieur pour parcourir leur contenu.

## 2. Les listes

Une liste permet de stocker plusieurs éléments dans une même variable.

Elle est créée avec des crochets [ ].

```python
nombres = [10, 20, 30, 40, 50]
```

Une liste peut contenir différents types de données :

```python
informations = ["Djah", 25, 1.85, True]
```

Cependant, dans la pratique, on regroupe généralement des données de même nature :

```python
notes = [12, 15, 17, 9, 14]
```

## 3. Accéder aux éléments d'une liste

Chaque élément possède une position, appelée indice.

Les indices commencent à 0.

```python
fruits = ["pomme", "banane", "orange", "mangue"]
```

| Élément | pomme | banane | orange | mangue |
| ------- | ----- | ------ | ------ | ------ |
| Indice  | 0     | 1      | 2      | 3      |


On peut accéder à un élément grâce à son indice :

```python
print(fruits[0])
```

```python
print(fruits[2])
```

Indices négatifs

Python permet également de partir de la fin :

```python
print(fruits[-1])
```

```python
print(fruits[-2])
```

## 4. Modifier une liste

Les listes sont modifiables.

On peut remplacer un élément :

```python
fruits = ["pomme", "banane", "orange"]

fruits[1] = "mangue"

print(fruits)
```

On peut également ajouter un élément avec append() :

```python
fruits.append("ananas")
```

```python
print(fruits)
```

Supprimer un élément avec remove() :

```python
fruits.remove("pomme")
print(fruits)
```

Ou supprimer un élément à partir de son indice avec pop() :

```python
fruits.pop(1)
```

```python
print(fruits)
```

## 5. La longueur d'une liste

La fonction len() permet de connaître le nombre d'éléments :

```python
notes = [12, 15, 17, 9, 14]

print(len(notes))
```

## 6. Les listes et les opérateurs

On peut concaténer deux listes avec + :

```python
liste1 = [1, 2, 3]
liste2 = [4, 5, 6]

resultat = liste1 + liste2
print(resultat)
```

On peut également répéter une liste :

```python
liste = [0] * 5
liste
```

L'opérateur **in** permet de vérifier si un élément appartient à une liste :

```python
fruits = ["pomme", "orange", "mangue"]
print("orange" in fruits)
```

## 7. Les tuples

Un tuple ressemble beaucoup à une liste, mais il est non modifiable.

Il est généralement créé avec des parenthèses :

```python
coordonnees = (10, 20)
```

On peut accéder à ses éléments comme avec une liste :

```python
print(coordonnees[0])
```

Mais on ne peut pas modifier un élément :

```python
# Le code suivant va provoquer une erreur.
coordonnees[0] = 50
```

## 8. Liste ou tuple ?

La différence fondamentale est donc :
| Liste                                   | Tuple                                       |
| --------------------------------------- | ------------------------------------------- |
| `[]`                                    | `()`                                        |
| Modifiable                              | Non modifiable                              |
| `list`                                  | `tuple`                                     |
| Adaptée aux données qui peuvent évoluer | Adapté aux données qui doivent rester fixes |


**Exemple**

Une liste peut être utilisée pour des notes :

notes = [12, 15, 17]

Parce qu'on peut ajouter ou modifier une note.

Un tuple peut être utilisé pour des coordonnées :

coordonnees = (9.5, 11.2)

ou pour représenter une date :

date_naissance = (2000, 10, 3)

## 9. Les dictionnaires

Un dictionnaire permet de stocker des données sous la forme :

clé → valeur

Il est créé avec des accolades {}.

```python
personne = {
    "nom": "Ali",
    "age": 25,
    "ville": "Moroni"
}
personne
```

Ici :

"nom"   → "Ali"  
"age"   → 25  
"ville" → "Moroni"  

La clé permet d'identifier la donnée.

## 10. Accéder aux valeurs d'un dictionnaire

Contrairement à une liste, on accède généralement aux données d'un dictionnaire grâce à leur clé.

```python
print(personne["nom"])
```

```python
print(personne["age"])
```

## 11. Modifier un dictionnaire

Les dictionnaires sont modifiables.

On peut modifier une valeur :

```python
personne["age"] = 26
```

On peut ajouter une nouvelle paire clé-valeur :

```python
personne["profession"] = "Enseignant"
```

On peut supprimer une donnée :

```python
del personne["ville"]
```

## 12. Vérifier l'existence d'une clé

L'opérateur in permet de vérifier si une clé existe :

```python
personne = {
    "nom": "Ali",
    "age": 25
}

print("nom" in personne)
```

On peut également utiliser get() :

```python
print(personne.get("nom"))
```

Si la clé n'existe pas, python renvoie **None**.

```python
print(personne.get("ville"))
```

Cela permet d'éviter une erreur lorsqu'on ne sait pas si une clé existe.

## 13. Dictionnaire avec plusieurs types de données

Les valeurs d'un dictionnaire peuvent être de types différents :

```python
etudiant = {
    "nom": "Ahmed",
    "age": 22,
    "moyenne": 14.5,
    "admis": True
}
```

On peut également avoir une liste comme valeur :

```python
etudiant = {
    "nom": "Ahmed",
    "notes": [12, 15, 17, 14]
}
```

On peut alors accéder à la liste :

```python
print(etudiant["notes"])
```

et à un élément de cette liste :

```python
print(etudiant["notes"][0])
```

**Cette combinaison de collections est extrêmement importante en Python.**


> À retenir

>Les trois structures étudiées répondent à des besoins différents :

>LISTE
→ plusieurs éléments modifiables  
→ [10, 20, 30]  

>TUPLE
→ plusieurs éléments non modifiables  
→ (10, 20, 30)  

>DICTIONNAIRE
→ données organisées sous forme : clé → valeur  
→ {"nom": "Ali", "age": 25}  

Le choix de la collection dépend donc de la nature des données et de la manière dont on souhaite les manipuler.
