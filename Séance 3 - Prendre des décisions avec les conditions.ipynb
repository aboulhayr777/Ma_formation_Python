## 1- Comparaisons
Avec **python**, on utilise les symboles suivants pour faire des comparaisons :

**_Remarque_** : _A chaque comparaison, il y aura une réponse de type bool **(True/False)**._

## 2- Structure conditionnelle `if`, `elif` et `else`:
**if** signifie **si**, **elif** signifie **sinon si** et **else** signifie **sinon**.

````python
age = 15

if age >= 18 :
    print("Vous êtes majeur.")
else : 
    print("Vous êtes mineur.")
````

````python
note = 14

if note >= 16:
    print("Très bien")
elif note >= 12:
    print("Bien")
else:
    print("Insuffisant")
````

## 3- Utilisation de la structure conditionnelle match/case
Python propose l'instruction **match/case** pour effectuer une sélection parmi plusieurs possibilités, en fonction de la valeur ou de la structure d'une donnée.  
Elle peut être vue comme une alternative plus lisisble à une succession de *if / elif / else* lorsqu'on cherche principalement à comparer **une même expression à plusieurs cas possibles**.

````python
jour = ""

match jour:
    case "lundi":
        # Instructions
        print("Début de semaine")
    case "vendredi":
        # Instructions
        print("Presque weekend")
    case _:
        # Cas par défaut
        print("Jour normal")
````

## 4- Utilisation de `and` et `or` avec les conditions
Les opérateurs logiques and et or

Jusqu'à présent, nous avons vu comment exécuter des instructions en fonction d'une condition :

````python
age = 20

if age >= 18:
    print("Majeur")
````

Mais dans un programme, il est fréquent de devoir vérifier plusieurs conditions en même temps.

Par exemple, pour autoriser une personne à accéder à un service, on peut vouloir vérifier qu'elle est majeure et qu'elle possède un compte. Python permet de combiner ces conditions grâce aux opérateurs logiques and et or.
___


1. L'opérateur and

L'opérateur and signifie « et ».

Il permet de vérifier que plusieurs conditions sont toutes vraies.

````python
age = 20
a_un_compte = True

if age >= 18 and a_un_compte:
    print("Accès autorisé")
````

Ici, l'accès est autorisé seulement si :

- age >= 18 est vrai ;
- et a_un_compte est vrai.

Si une seule des deux conditions est fausse, l'ensemble de la condition est faux.

Exemple :

````python
age = 16
a_un_compte = True

if age >= 18 and a_un_compte:
    print("Accès autorisé")
else:
    print("Accès refusé")
````

2. L'opérateur or

L'opérateur or signifie « ou ».

Il permet de vérifier qu'au moins une des conditions est vraie.

````python
jour = "samedi"

if jour == "samedi" or jour == "dimanche":
    print("C'est le week-end")
````

### Tableau de vérité
Pour bien comprendre la différence :

| Condition A | Condition B | `A and B` | `A or B` |
| ----------- | ----------- | --------- | -------- |
| Vrai        | Vrai        | Vrai      | Vrai     |
| Vrai        | Faux        | Faux      | Vrai     |
| Faux        | Vrai        | Faux      | Vrai     |
| Faux        | Faux        | Faux      | Faux     |
