# La boucle for en Python

## 1. Introduction

Dans un programme, il est fréquent de devoir répéter plusieurs fois une même série d'instructions.

Par exemple, afficher les nombres de 1 à 10 :

```python
print(1)
print(2)
print(3)
# ...
print(10)
```

Cette méthode devient rapidement longue et peu pratique.

La boucle for permet d'exécuter plusieurs fois un bloc d'instructions en parcourant successivement les éléments d'une séquence.

On peut ainsi parcourir :

- une liste ;
- une chaîne de caractères ;
- un tuple ;
- un dictionnaire ;
- une suite de nombres générée avec range() ;
etc.

## 2. Syntaxe de la boucle for

La syntaxe générale est :

```python
for variable in sequence:
    instruction
```

```python
for nombre in [1, 2, 3, 4, 5]:
    print(nombre)
```

Comment cela fonctionne ?

À chaque répétition, la variable nombre prend successivement une valeur de la liste :  
nombre = 1  
nombre = 2  
nombre = 3  
nombre = 4  
nombre = 5    

## 3. Parcourir une chaîne de caractères

Une chaîne de caractères est également une séquence.

On peut donc parcourir ses caractères avec for :

```python
mot = "Python"

for caractere in mot:
    print(caractere)
```

Cela permet notamment de réaliser des traitements caractère par caractère

## 4. Utiliser range()

Très souvent, on souhaite répéter une instruction un certain nombre de fois.

La fonction range() permet de générer une suite de nombres.

```python
for i in range(5):
    print(i)
```

range(5) commence à 0 et s'arrête avant 5.

```python
# Donc
list(range(5))
```

## 5. range() avec un début et une fin

On peut préciser le début et la fin :

```python
for i in range(1, 6):
    print(i)
```

La syntaxe est : range(debut, fin)  
La valeur fin n'est jamais incluse.

## 6. Ajouter un pas

On peut également définir le pas :

```python
for i in range(0, 11, 2):
    print(i)
```

La syntaxe est : range(debut, fin, pas)  
La valeur fin n'est jamais incluse.

On peut également utiliser un pas négatif :

```python
for i in range(10, 0, -1):
    print(i)
```

## 7. Utiliser for avec une condition

Une boucle for peut être combinée avec une structure conditionnelle.

Par exemple, afficher uniquement les nombres pairs entre 1 et 20 :

```python
for nombre in range(1, 21):
    if nombre % 2 == 0:
        print(nombre)
```

On combine donc ici deux notions :  
for → parcourir les nombres  
if  → sélectionner ceux qui nous intéressent  
C'est une combinaison extrêmement fréquente en programmation.

## Applications
### Application 1 — Table de multiplication

Écrire un programme qui demande à l'utilisateur un nombre entier puis affiche sa table de multiplication de 1 à 10.

```python
nombre = int(input("Saisir in nombre entier"))

for i in range(1, 10):
    print(f"{nombre} * {i} = {nombre*i}")
```

### Application 2 — Somme des nombres

Écrire un programme qui calcule la somme des nombres entiers de 1 à 100.


```python
somme = 0
for i in range(0, 101):
    somme += i

print(f"Somme des cents premiers est {somme}")
```


### Application 3 — Compter les nombres pairs

Écrire un programme qui parcourt les nombres de 1 à 100 et compte combien d'entre eux sont pairs.

Le programme doit afficher le nombre total de nombres pairs.

```python
nbr_pair = 0
for i in range(1, 101):
    if i%2 == 0:
        nbr_pair += 1

print(f"Nombre pair = {nbr_pair}")
```

### Application 4 — Analyse d'une liste

On considère la liste suivante :  
notes = [12, 15, 8, 17, 10, 14, 6, 19].  

Écrire un programme qui parcourt cette liste et affiche uniquement les notes supérieures ou égales à 10.  

Puis modifier le programme afin de calculer :

- le nombre de notes supérieures ou égales à 10 ;
- le nombre de notes inférieures à 10 ;
- la moyenne des notes.

```python
notes = [12, 15, 8, 17, 10, 14, 6, 19]
note_less_10 = []
note_under_10 = []

somme_note = 0
for note in notes:
    somme_note += note
    if note >= 10:
        note_under_10.append(note)
    else:
        note_less_10.append(note)

moyenne_note = somme_note/len(notes)
print(f"Les notes supérieures à 10 : {note_under_10}")
print(f"Les notes inférieures à 10 : {note_less_10}")
print(f"Moyenne = {moyenne_note}")
```

### Application 5 — Recherche d'un élément

On considère :

nombres = [12, 7, 25, 4, 18, 31, 9]  

Écrire un programme qui demande un nombre à l'utilisateur et recherche ce nombre dans la liste.  

Si le nombre est présent, afficher :  

Le nombre a été trouvé.  

Sinon :  

Le nombre n'a pas été trouvé.  

```python
nombres = [12, 7, 25, 4, 18, 31, 9]
nombre = int(input("Saisir un nombre : "))
msg = "Le nombre n'a pas été trouvé"

for nbr in nombres :
    if nbr == nombre :
        msg = "Le nombre est trouvé."

print(msg)
```


### Application 6 — Analyse d'un mot

Écrire un programme qui demande à l'utilisateur de saisir un mot et compte le nombre de voyelles qu'il contient.

Par exemple :

Entrée : programmation
Sortie : 5 voyelles

Cet exercice est particulièrement intéressant parce qu'il oblige à combiner :

for
if

et l'opérateur in :

if caractere in "aeiou":

```python
voyelles = "aeiou"
mot = input("Donner un mot : ")
nbr_voyelle = 0
for lettre in mot:
    if lettre in voyelles : 
        nbr_voyelle += 1
        
print(f"Le nombre de voyelles dans le mot : {nbr_voyelle} voyelles.")
```

### 8. break : arrêter une boucle

L'instruction **break** permet d'interrompre immédiatement une boucle.

Par exemple :

```python
for nombre in range(1, 11):
    if nombre == 5:
        break

    print(nombre)
```

Lorsque nombre devient égal à 5, break arrête la boucle.

**Application**

Reprendre l'exercice de recherche d'un élément dans une liste et utiliser break pour arrêter la recherche dès que l'élément est trouvé.

## 9. continue : passer à l'itération suivante

**continue** permet de sauter le reste du bloc actuel et de passer directement à l'itération suivante.

```python
for nombre in range(1, 11):
    if nombre % 2 == 0:
        continue

    print(nombre)
```

Lorsque le nombre est pair, continue permet de passer directement au nombre suivant.

## 10. Les boucles for imbriquées

Une boucle peut contenir une autre boucle.

Par exemple :

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

La boucle intérieure est exécutée entièrement pour chaque valeur de la boucle extérieure.

Cette technique est notamment utilisée pour :

- parcourir des tableaux à deux dimensions ;
- créer des tableaux ;
- travailler avec des coordonnées ;
- générer des combinaisons ;
- construire certaines figures avec des caractères.

**Application**

Écrire un programme qui affiche un carré de * de taille 5 × 5. Puis modifier le programme pour obtenir un triangle.

```python
for i in range(5):
    print("*****")
```

```python
for i in range(5):
    print("*"*(i+1))
```
